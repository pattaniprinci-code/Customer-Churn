```python

# Use a subset for hyperparameter tuning
n_search = 15000

rng = np.random.RandomState(42)
idx = rng.choice(len(X_train_scal), n_search, replace=False)

X_search = X_train_scal[idx]
y_search = y_train.iloc[idx] if hasattr(y_train, "iloc") else y_train[idx]
```

```python
search = RandomizedSearchCV(
        estimator=SVR(),
        cv=5,
        param_distributions=svr_params,
        n_iter=2,
        random_state=42,
        scoring="r2"
    )
```

```python
print("Best Score:", search.best_score_)
print("Best Parameters:", search.best_params_)
```
